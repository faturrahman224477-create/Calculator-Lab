# KODE REGRESI PAYAKUMBUH

```python
# Import Library
import numpy as np
import pandas as pd
import statsmodels.api as sm

# Membaca dataset dari URL GitHub Anda
url = "https://raw.githubusercontent.com/Alam-A99/DATAAKSI/main/dataset_fiskal_97_hipoFull.csv"
df = pd.read_csv(url)

# --- MEMFILTER KHUSUS KOTA PAYAKUMBUH ---
# Menggunakan .str.contains atau pencarian agar aman dari perbedaan penulisan huruf besar/kecil
df_payakumbuh = df[df['Kota'].astype(str).str.contains('Payakumbuh', case=False, na=False)]

print(f"Jumlah data untuk Kota Payakumbuh: {len(df_payakumbuh)}")

# Memisahkan variabel independen (X1, X2) dan dependen (y)
X = df_payakumbuh[['bm_pendidikan', 'PAD']]  # X1 dan X2
y = df_payakumbuh['DAU_Earmark']            # Y (Variabel Target)

X = sm.add_constant(X) # Menambahkan kolom konstanta untuk intercept (Beta 0)

model = sm.OLS(y, X) # Membangun model regresi linear berganda
results = model.fit()

print(results.summary()) # Menampilkan hasil ringkasan analisis regresi
```

Berikut adalah interpretasi hasil regresi untuk **Kota Payakumbuh**

* **Persamaan Regresi:**

$$\text{DAU\_Earmark} = 7.5840 + 0.0424(\text{bm\_pendidikan}) + 0.1293(\text{PAD})$$


* **Koefisien Determinasi ($R^2$):**
Nilai $R$-squared sebesar $0.524$ menunjukkan bahwa kemampuan variabel independen (`bm_pendidikan` dan `PAD`) dalam menjelaskan variasi variabel dependen (`DAU_Earmark`) pada Kota Payakumbuh adalah sebesar $52.4\%$.
* **Uji Simultan (F-statistic):**
Nilai *Prob (F-statistic)* sebesar $0.0513$ menunjukkan bahwa secara bersama-sama variabel independen memiliki pengaruh yang mendekati signifikan terhadap variabel dependen pada tingkat keyakinan $\alpha = 0.05$.
* **Pengaruh Belanja Modal Pendidikan (`bm_pendidikan`):**
Memiliki koefisien $0.0424$ dengan nilai $p$-value $0.050$. Variabel ini dinyatakan **berpengaruh signifikan** terhadap `DAU_Earmark`. Setiap kenaikan 1 satuan pada belanja modal pendidikan akan menaikkan DAU Earmark sebesar $0.0424$ satuan.
* **Pengaruh Pendapatan Asli Daerah (`PAD`):**
Memiliki koefisien $0.1293$ dengan nilai $p$-value $0.253$. Variabel ini dinyatakan **tidak berpengaruh signifikan** terhadap `DAU_Earmark` karena nilai $p$-value lebih besar dari $0.05$.
